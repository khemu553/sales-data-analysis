```python
sales.drop_duplicates(inplace=True)
```

check missing values and fill

```python
sales.isnull().sum()
```

```python
sales['Region']=sales['Region'].fillna('unknown')
sales['Discount']=sales['Discount'].fillna(0)
sales.isnull().sum()
```

new column month and year

```python
sales['Order_Date']=pd.to_datetime(sales['Order_Date'])
```

```python
sales["Month"]=sales["Order_Date"].dt.month
sales["Year"]=sales["Order_Date"].dt.year
```

```python
sales
```

profit_margin

```python
sales["Profit_margin"]=(sales['Profit']/sales['Sales']*100).round(2)
sales
```

groupby category -wise sales,region- wise profit ,top 10 customer


```python
sales.groupby('Category')['Sales'].sum().sort_values(ascending=False)
```

```python
sales.groupby('Region')['Profit'].sum().sort_values(ascending=False)
```

```python
sales.groupby('Customer_ID')['Sales'].sum().sort_values(ascending=False).head(10)
```
